## Q03: Binary Search
<a id="q03-binarysearch"></a>

Search for a key in a sorted array using standard iterative binary search.


```python
# Iterative Binary Search

arr = list(map(int, input("Enter sorted elements (space-separated): ").split()))
key = int(input("Enter element to search: "))

low = 0
high = len(arr) - 1
found = False

while low <= high:
    mid = (low + high) // 2
    if arr[mid] == key:
        print(f"Element {key} found at index {mid}")
        found = True
        break
    elif arr[mid] < key:
        low = mid + 1
    else:
        high = mid - 1

if not found:
    print(f"Element {key} not found in array")
```

## Q07: Hash Table with Linear Probing
<a id="q07-linearprobing"></a>

Resolve hash table collisions using open addressing with Linear Probing: `(index + 1) % SIZE`.


```python
# Hash Table with Linear Probing

SIZE = 10
table = [-1] * SIZE

n = int(input("Enter number of keys: "))
keys = list(map(int, input(f"Enter {n} keys: ").split()))

for key in keys:
    pos = key % SIZE
    # Look for next empty slot
    while table[pos] != -1:
        pos = (pos + 1) % SIZE
    table[pos] = key

print("\nHash Table (Linear Probing):")
for i in range(SIZE):
    print(f"Slot [{i}]: {table[i]}")
```

## Q08: Hash Table with Quadratic Probing
<a id="q08-quadraticprobing"></a>

Resolve hash table collisions using Quadratic Probing: `(hash + j*j) % SIZE`.


```python
# Hash Table with Quadratic Probing

SIZE = 11
table = [-1] * SIZE

n = int(input("Enter number of keys: "))
keys = list(map(int, input(f"Enter {n} keys: ").split()))

for key in keys:
    h = key % SIZE
    pos = h
    j = 1
    # Probe quadratically: h + 1^2, h + 2^2, ...
    while table[pos] != -1:
        pos = (h + j * j) % SIZE
        j += 1
    table[pos] = key

print("\nHash Table (Quadratic Probing):")
for i in range(SIZE):
    print(f"Slot [{i}]: {table[i]}")
```

## Q10: Binary Search (Function Form)
<a id="q10-binarysearch-2"></a>

Modular binary search implemented as a reusable function returning the found index or -1.


```python
# Binary Search as a function

def binary_search(arr, key):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1
    return -1

arr = list(map(int, input("Enter sorted elements: ").split()))
key = int(input("Enter key to search: "))

result = binary_search(arr, key)
if result != -1:
    print(f"Found at index: {result}")
else:
    print("Not found in array")
```

## Q12: Binary Tree — Total Node Count
<a id="q12-tree-nodecount"></a>

Construct a binary tree and recursively compute the total count of nodes.


```python
# Binary Tree - Count Total Nodes

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def count_nodes(root):
    if root is None:
        return 0
    return 1 + count_nodes(root.left) + count_nodes(root.right)

# Build sample tree:
#       1
#      / \
#     2   3
#    / \
#   4   5
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

print(f"Total number of nodes in tree: {count_nodes(root)}")
```

## Q13: Binary Tree — Leaf Node Count

Traverse a binary tree to count leaf nodes (nodes with no children).


```python
# Binary Tree - Count Leaf Nodes

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def count_leaves(root):
    if root is None:
        return 0
    if root.left is None and root.right is None:
        return 1
    return count_leaves(root.left) + count_leaves(root.right)

# Sample tree
root = TreeNode(10)
root.left = TreeNode(20)
root.right = TreeNode(30)
root.left.left = TreeNode(40)
root.left.right = TreeNode(50)

print(f"Total leaf nodes: {count_leaves(root)}")
```

## Q15: Binary Tree — Height
<a id="q15-tree-height"></a>

Compute the height (maximum path length from root to leaf) of a binary tree.


```python
# Binary Tree - Calculate Height

class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def tree_height(root):
    if root is None:
        return 0
    return 1 + max(tree_height(root.left), tree_height(root.right))

# Sample tree
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)

print(f"Height of tree: {tree_height(root)}")
```

## Q22: Graph — Breadth First Search (BFS)
<a id="q22-bfs"></a>

Perform Breadth First Search graph traversal using a queue.


```python
# Breadth First Search (BFS)

from collections import deque

def bfs(graph, start):
    visited = set([start])
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return order

# Sample graph
graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 4],
    3: [1],
    4: [1, 2]
}

print(f"BFS Traversal starting from vertex 0: {bfs(graph, 0)}")
```

## Q23: Graph — Depth First Search (DFS)
<a id="q23-dfs"></a>

Perform Depth First Search graph traversal using recursion.


```python
# Depth First Search (DFS)

def dfs(graph, start, visited=None, order=None):
    if visited is None:
        visited = set()
        order = []
    visited.add(start)
    order.append(start)

    for neighbor in graph.get(start, []):
        if neighbor not in visited:
            dfs(graph, neighbor, visited, order)
    return order

# Sample graph
graph = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0],
    3: [1],
    4: [1]
}

print(f"DFS Traversal starting from vertex 0: {dfs(graph, 0)}")
```

## Q24: Bellman-Ford Shortest Path Algorithm
<a id="q24-bellmanford"></a>

Compute shortest distances from a source vertex to all vertices in a weighted graph.


```python
# Bellman-Ford Shortest Path

def bellman_ford(vertices, edges, src):
    dist = {v: float('inf') for v in range(vertices)}
    dist[src] = 0

    # Relax edges V - 1 times
    for _ in range(vertices - 1):
        for u, v, w in edges:
            if dist[u] != float('inf') and dist[u] + w < dist[v]:
                dist[v] = dist[u] + w

    # Check for negative weight cycles
    for u, v, w in edges:
        if dist[u] != float('inf') and dist[u] + w < dist[v]:
            print("Graph contains negative weight cycle!")
            return None

    return dist

# Sample: (u, v, weight)
edges = [
    (0, 1, 4),
    (0, 2, 2),
    (1, 2, 3),
    (2, 1, 1),
    (1, 3, 2),
    (2, 3, 5)
]
v_count = 4
src = 0
distances = bellman_ford(v_count, edges, src)
print(f"Shortest distances from source {src}:")
for v in distances:
    print(f"Vertex {v}: Distance = {distances[v]}")
```

## Q25: Dijkstra Shortest Path Algorithm
<a id="q25-dijkstra"></a>

Find single-source shortest paths in a non-negative weighted graph.


```python
# Dijkstra Shortest Path Algorithm

def dijkstra(graph, v_count, src):
    dist = [float('inf')] * v_count
    visited = [False] * v_count
    dist[src] = 0

    for _ in range(v_count):
        # Pick vertex with min distance
        u = -1
        for i in range(v_count):
            if not visited[i] and (u == -1 or dist[i] < dist[u]):
                u = i

        if dist[u] == float('inf'):
            break

        visited[u] = True

        for v, weight in graph[u]:
            if not visited[v] and dist[u] + weight < dist[v]:
                dist[v] = dist[u] + weight

    return dist

# Sample graph as adjacency list: (neighbor, weight)
graph = {
    0: [(1, 4), (2, 1)],
    1: [(3, 1)],
    2: [(1, 2), (3, 5)],
    3: []
}

res = dijkstra(graph, 4, 0)
for i, d in enumerate(res):
    print(f"Shortest path from 0 to {i} = {d}")
```

## Q26: Prim's Minimum Spanning Tree
<a id="q26-prims"></a>

Find the Minimum Spanning Tree (MST) using Prim's greedy algorithm.


```python
# Prim's MST Algorithm

def prim_mst(graph, v_count):
    key = [float('inf')] * v_count
    parent = [-1] * v_count
    in_mst = [False] * v_count

    key[0] = 0

    for _ in range(v_count):
        u = -1
        for i in range(v_count):
            if not in_mst[i] and (u == -1 or key[i] < key[u]):
                u = i

        in_mst[u] = True

        for v in range(v_count):
            if graph[u][v] != 0 and not in_mst[v] and graph[u][v] < key[v]:
                key[v] = graph[u][v]
                parent[v] = u

    print("MST Edges:")
    total_cost = 0
    for i in range(1, v_count):
        print(f"Edge {parent[i]} - {i} (weight: {graph[i][parent[i]]})")
        total_cost += graph[i][parent[i]]
    print(f"Total MST Weight = {total_cost}")

# Sample graph matrix
graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]
prim_mst(graph, 5)
```

## Q27: Kruskal's Minimum Spanning Tree
<a id="q27-kruskal"></a>

Find MST by sorting edges and avoiding cycles using Disjoint Set Union.


```python
# Kruskal's MST Algorithm

class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))

    def find(self, i):
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i])
        return self.parent[i]

    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)
        if root_i != root_j:
            self.parent[root_i] = root_j
            return True
        return False

def kruskal(v_count, edges):
    # Sort edges by weight
    edges.sort(key=lambda x: x[2])
    ds = DisjointSet(v_count)
    mst = []
    total_weight = 0

    for u, v, w in edges:
        if ds.union(u, v):
            mst.append((u, v, w))
            total_weight += w

    print("MST Edges (Kruskal):")
    for u, v, w in mst:
        print(f"{u} - {v} : weight {w}")
    print(f"Total Weight: {total_weight}")

edges = [
    (0, 1, 10), (0, 2, 6), (0, 3, 5),
    (1, 3, 15), (2, 3, 4)
]
kruskal(4, edges)
```

## Q29: 0/1 Knapsack — Dynamic Programming (Tabulation)
<a id="q29-tabulation"></a>

Solve the 0/1 Knapsack problem using a 2D dynamic programming table.


```python
# 0/1 Knapsack using Dynamic Programming

values = list(map(int, input("Enter item values (e.g. 60 100 120): ").split()))
weights = list(map(int, input("Enter item weights (e.g. 10 20 30): ").split()))
capacity = int(input("Enter maximum knapsack capacity: "))
n = len(values)

dp = [[0] * (capacity + 1) for _ in range(n + 1)]

for i in range(1, n + 1):
    for w in range(1, capacity + 1):
        if weights[i - 1] <= w:
            dp[i][w] = max(dp[i - 1][w], values[i - 1] + dp[i - 1][w - weights[i - 1]])
        else:
            dp[i][w] = dp[i - 1][w]

print(f"Maximum knapsack value achievable: {dp[n][capacity]}")
```

## Q35: Price Sorting — 3 Methods Comparison
<a id="q35-pricesorting-3methods"></a>

Compare Bubble Sort, Selection Sort, and Insertion Sort on a list of item prices.


```python
# Price Sorting Comparison: Bubble, Selection, Insertion Sort

def bubble_sort(arr):
    a = arr.copy()
    for i in range(len(a) - 1):
        for j in range(len(a) - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

def selection_sort(arr):
    a = arr.copy()
    for i in range(len(a)):
        min_idx = i
        for j in range(i + 1, len(a)):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]
    return a

def insertion_sort(arr):
    a = arr.copy()
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a

prices = list(map(float, input("Enter prices: ").split()))
print("Bubble Sort:   ", bubble_sort(prices))
print("Selection Sort:", selection_sort(prices))
print("Insertion Sort:", insertion_sort(prices))
```

## Q37: Ternary Search
<a id="q37-ternarysearch"></a>

Search for a key in a sorted array by dividing the search interval into three parts.


```python
# Ternary Search

def ternary_search(arr, low, high, key):
    if low > high:
        return -1

    mid1 = low + (high - low) // 3
    mid2 = high - (high - low) // 3

    if arr[mid1] == key:
        return mid1
    if arr[mid2] == key:
        return mid2

    if key < arr[mid1]:
        return ternary_search(arr, low, mid1 - 1, key)
    elif key > arr[mid2]:
        return ternary_search(arr, mid2 + 1, high, key)
    else:
        return ternary_search(arr, mid1 + 1, mid2 - 1, key)

arr = list(map(int, input("Enter sorted elements: ").split()))
key = int(input("Enter key to search: "))

idx = ternary_search(arr, 0, len(arr) - 1, key)
if idx != -1:
    print(f"Found {key} at index {idx}")
else:
    print(f"{key} not found")
```

## Q38: Quick Sort
<a id="q38-quicksort"></a>

Sort an array using the divide-and-conquer Quick Sort algorithm.


```python
# Quick Sort Algorithm

def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)

arr = list(map(int, input("Enter elements to sort: ").split()))
sorted_arr = quick_sort(arr)
print(f"Sorted array: {sorted_arr}")
```

## Q39: Bubble Sort

Classic Bubble Sort algorithm with step-by-step adjacent element swapping.


```python
# Bubble Sort

arr = list(map(int, input("Enter elements to sort: ").split()))
n = len(arr)

for i in range(n - 1):
    swapped = False
    for j in range(n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
            swapped = True
    if not swapped:
        break

print(f"Sorted array: {arr}")
```

## Q41: Insertion Sort

Insertion Sort algorithm: builds the sorted list one element at a time.


```python
# Insertion Sort

arr = list(map(int, input("Enter elements to sort: ").split()))

for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1
    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1
    arr[j + 1] = key

print(f"Sorted array: {arr}")
```

## Q52: Coin Change (Greedy vs Dynamic Programming)

Compare greedy coin change against dynamic programming for minimum coins required.


```python
# Coin Change - Greedy vs DP

coins = [1, 5, 10, 25]
amount = int(input("Enter amount to make change for: "))

# 1. Greedy approach
greedy_coins = []
rem = amount
for c in reversed(coins):
    while rem >= c:
        greedy_coins.append(c)
        rem -= c

# 2. Dynamic Programming approach (minimum coins)
dp = [float('inf')] * (amount + 1)
dp[0] = 0

for a in range(1, amount + 1):
    for c in coins:
        if a - c >= 0:
            dp[a] = min(dp[a], dp[a - c] + 1)

print(f"Greedy Coins: {greedy_coins} (Total: {len(greedy_coins)})")
print(f"DP Minimum Coins count: {dp[amount]}")
```

## Q57: Binary Search in Descending Order Array
<a id="q57-binarysearch-descending"></a>

Given an array sorted in descending order and a target value, find the index of the target (or return -1 if absent).


```python
# Binary Search in Descending Order

arr = list(map(int, input("Enter descending sorted array (e.g. 20 15 11 8 4 1): ").split()))
target = int(input("Enter target value to search: "))

low = 0
high = len(arr) - 1
found_index = -1

while low <= high:
    mid = (low + high) // 2
    if arr[mid] == target:
        found_index = mid
        break
    elif arr[mid] < target:
        # In descending order, larger elements lie to the left
        high = mid - 1
    else:
        # Smaller elements lie to the right
        low = mid + 1

if found_index != -1:
    print(f"Target {target} found at index: {found_index}")
else:
    print(f"Target {target} not found (-1)")
```

## Q59: Valid Parentheses Using Stack
<a id="q59-validparentheses-stack"></a>

Given a string containing brackets '(', ')', '{', '}', '[', ']', determine whether the brackets are correctly balanced and nested.


```python
# Valid Parentheses Checker using Stack

def is_valid_brackets(s):
    stack = []
    bracket_pairs = {')': '(', '}': '{', ']': '['}

    for char in s:
        if char in bracket_pairs.values():
            stack.append(char)
        elif char in bracket_pairs:
            if not stack or stack.pop() != bracket_pairs[char]:
                return False

    return len(stack) == 0

s = input("Enter bracket string (e.g. {[()]} or ()[]{}): ").strip()

if is_valid_brackets(s):
    print("Result: BALANCED (Valid Parentheses)")
else:
    print("Result: NOT BALANCED (Invalid Parentheses)")
```
